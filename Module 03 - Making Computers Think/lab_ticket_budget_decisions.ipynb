Hey! Please write a simple, Python script for a "Purchase Advisor" tool to help users decide if they can afford buying something.

Here are the specific requirements for the code:

1. Inputs needed:
   - Monthly income
   - Total current savings
   - Price of the item they want to buy
   (Make sure these are all floating-point numbers)

2. Input Validation:
   - Check if any input is 0 or negative. If so, show an error message and exit early.

3. Calculations:
   - Item price as a percentage of monthly income
   - Remaining savings after buying the item

4. Decision Logic (it must trigger different outcomes based on these exact rules):
   - Option A (Smart Buy): Item price is <= 10% of savings AND <= 20% of monthly income.
   - Option B (Caution): Item price is between 10%-30% of savings OR between 20%-50% of monthly income.
   - Option C (High Risk): Item price is > 30% of savings OR > 50% of monthly income.

5. Output:
   - Display an easy-to-read summary showing the user's income, savings, item price, the calculated percentages, and the final recommendation.

Please give me only the complete, ready-to-run Python code without asking any follow-up questions.

```python
def purchase_advisor():
    print("=" * 40)
    print("      SMART PURCHASE ADVISOR TOOL      ")
    print("=" * 40)

    try:
        income = float(input("Enter your Monthly Income ($): "))
        savings = float(input("Enter your Total Savings ($): "))
        price = float(input("Enter the Price of the Item ($): "))

        if income <= 0 or savings <= 0 or price <= 0:
            print("\n[ERROR] All monetary amounts must be greater than zero. Please try again.")
            return
    except ValueError:
        print("\n[ERROR] Invalid input. Please enter valid numbers.")
        return

    price_to_income_pct = (price / income) * 100
    price_to_savings_pct = (price / savings) * 100
    remaining_savings = savings - price

    print("\n" + "-" * 40)
    print("          FINANCIAL SUMMARY           ")
    print("-" * 40)
    print(f"Monthly Income    : ${income:,.2f}")
    print(f"Current Savings   : ${savings:,.2f}")
    print(f"Item Price        : ${price:,.2f}")
    print(f"Price vs Income   : {price_to_income_pct:.1f}%")
    print(f"Price vs Savings  : {price_to_savings_pct:.1f}%")
    print("-" * 40)

    if price_to_savings_pct <= 10 and price_to_income_pct <= 20:
        print("DECISION: 🟢 SMART BUY")
        print("Advice   : Go for it! This purchase fits comfortably within your budget.")
        print(f"Impact   : You will still have ${remaining_savings:,.2f} in savings.")
    elif price_to_savings_pct > 30 or price_to_income_pct > 50:
        print("DECISION: 🔴 HIGH RISK - DO NOT BUY")
        print("Advice   : This item is too expensive relative to your current wealth.")
        print("           Buying this will severely weaken your financial safety net.")
    else:
        print("DECISION: 🟡 CAUTION - THINK TWICE")
        print("Advice   : This is a significant expense. Consider waiting 30 days.")
        print(f"Impact   : Your savings will drop to ${remaining_savings:,.2f}.")

    print("=" * 40)

if __name__ == "__main__":
    purchase_advisor()
```
